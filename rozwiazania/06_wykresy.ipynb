# Rozwiązania — lekcja 6: Wykresy, które odpowiadają na pytania

To przykładowe odpowiedzi. Inny poprawny sposób jest równie dobry. Każde rozwiązanie uruchom po komórce przygotowującej dane poniżej.


```python
from pathlib import Path

# Notebook działa po uruchomieniu z katalogu kursu lub z jego podkatalogu.
KATALOG_KURSU = next(
    katalog for katalog in [Path.cwd(), *Path.cwd().parents]
    if (katalog / "dane" / "sprzedaz_kawiarni.csv").exists()
)
DANE = KATALOG_KURSU / "dane"
WYNIKI = KATALOG_KURSU / "wyniki"
WYNIKI.mkdir(exist_ok=True)

import pandas as pd
import matplotlib.pyplot as plt

sprzedaz = pd.read_csv(DANE / "sprzedaz_kawiarni_gotowa.csv", parse_dates=["data"])
sprzedaz.head()
sprzedaz["przychod"] = sprzedaz["liczba_sztuk"] * sprzedaz["cena_jednostkowa"]
```

## Porównaj kanały

Zrób wykres słupkowy przychodu według kanału. Dodaj tytuł i jednostkę, a pod wykresem zapisz wniosek.


```python
kanaly = sprzedaz.groupby("kanal")["przychod"].sum()
ax = kanaly.plot.bar(rot=0, color="#387c9b", figsize=(6, 4))
ax.set(title="Przychód według kanału sprzedaży", xlabel="Kanał", ylabel="Przychód [zł]")
ax.set_ylim(bottom=0)
plt.tight_layout()
plt.show()
```

**Interpretacja / kontrola:** Dokładnie dwa słupki; ich wysokości zgadzają się z tabelą z lekcji 5. Wniosek dotyczy zarejestrowanego przychodu po czyszczeniu.


## Zapisz histogram

Zrób histogram wartości pozycji sprzedaży i zapisz go do `wyniki/histogram_przychodu.png`. Zmień liczbę przedziałów na 6, potem 15. Co się zmieniło?


```python
ax = sprzedaz["przychod"].plot.hist(bins=15, figsize=(7, 4), edgecolor="white")
ax.set(title="Rozkład wartości pozycji sprzedaży", xlabel="Wartość [zł]", ylabel="Liczba pozycji")
fig = ax.get_figure()
fig.savefig(WYNIKI / "histogram_przychodu.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Interpretacja / kontrola:** Plik PNG powstaje w `wyniki/`. Zmiana liczby przedziałów zmienia szczegółowość obrazu, ale nie zmienia danych.
