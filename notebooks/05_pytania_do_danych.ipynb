## 1. Ile wynosi wartość sprzedaży?

Działanie na dwóch kolumnach dopasowuje wartości z każdego wiersza. Nie potrzebujemy pętli. Kwoty są fikcyjnymi wartościami sprzedaży w zł; nie analizujemy podatków ani kosztów.


```python
sprzedaz["przychod"] = sprzedaz["liczba_sztuk"] * sprzedaz["cena_jednostkowa"]
display(sprzedaz[["produkt", "liczba_sztuk", "cena_jednostkowa", "przychod"]].head())
print("Łączny przychód:", sprzedaz["przychod"].sum(), "zł")
```

## 2. Co znaczy „najpopularniejszy”?

`value_counts()` liczy wiersze, a `sum()` ilości. Jedna pozycja może zawierać kilka sztuk. Zawsze nazwij miarę.


```python
print("Liczba pozycji sprzedaży:")
print(sprzedaz["produkt"].value_counts())
print("Sprzedane sztuki:")
print(sprzedaz.groupby("produkt")["liczba_sztuk"].sum().sort_values(ascending=False))
```

## 3. Kilka miar w jednej tabeli

`.agg()` nadaje nazwy wynikom. Zapis `("kolumna", "działanie")` określa, którą kolumnę i jak podsumować. `.reset_index()` przywraca kolumnę grupującą do zwykłych kolumn tabeli.


```python
sprzedaz["przychod"] = sprzedaz["liczba_sztuk"] * sprzedaz["cena_jednostkowa"]
podsumowanie = sprzedaz.groupby("kategoria").agg(
    pozycje=("id_transakcji", "count"),
    sztuki=("liczba_sztuk", "sum"),
    przychod_zl=("przychod", "sum"),
    srednia_wartosc_pozycji_zl=("przychod", "mean"),
).reset_index()
podsumowanie
```

Średnia wartość **pozycji sprzedaży** nie jest średnią wartością paragonu ani średnim wydatkiem klienta. Do takich obliczeń potrzebowalibyśmy innych identyfikatorów.
