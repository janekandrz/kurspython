## Pytanie na dziś

Co naprawdę opisuje plik sprzedaży kawiarni i czy można mu od razu zaufać?

**CSV** to plik tekstowy z tabelą. **pandas** to biblioteka do pracy z danymi. `import pandas as pd` udostępnia ją pod krótką nazwą `pd`. **DataFrame** to tabela, a **Series** to pojedyncza kolumna.

Jedna obserwacja w naszym pliku to **pozycja sprzedaży produktu**. Wiersz może dotyczyć kilku sztuk. `id_transakcji` identyfikuje pozycję; dane nie zawierają wspólnego identyfikatora paragonu ani liczby klientów.

Komórka poniżej znajduje folder kursu. To kod pomocniczy — uruchom go; nie musisz go teraz zapamiętywać.


```python
from pathlib import Path

# Notebook działa po uruchomieniu z katalogu kursu lub z jego podkatalogu.
KATALOG_KURSU = next(
    katalog for katalog in [Path.cwd(), *Path.cwd().parents]
    if (katalog / "dane" / "sprzedaz_kawiarni.csv").exists()
)
DANE = KATALOG_KURSU / "dane"
WYNIKI = KATALOG_KURSU / "wyniki"
WYNIKI.mkdir(exist_ok=True)

import pandas as pd
sprzedaz = pd.read_csv(DANE / "sprzedaz_kawiarni.csv")
```
