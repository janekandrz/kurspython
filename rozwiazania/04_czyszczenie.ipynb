# Rozwiązania — lekcja 4: Wybieranie i czyszczenie danych

To przykładowe odpowiedzi. Inny poprawny sposób jest równie dobry. Każde rozwiązanie uruchom po komórce przygotowującej dane poniżej.


```python
from pathlib import Path

# Notebook działa po uruchomieniu z katalogu kursu lub z jego podkatalogu.
KATALOG_KURSU = next(
    katalog for katalog in [Path.cwd(), *Path.cwd().parents]
    if (katalog / "dane" / "sprzedaz_kawiarni.csv").exists()
)
DANE = KATALOG_KURSU / "dane"
WYNIKI = KATALOG_KURSU / "wyniki"
WYNIKI.mkdir(exist_ok=True)

import pandas as pd
sprzedaz = pd.read_csv(DANE / "sprzedaz_kawiarni_gotowa.csv", parse_dates=["data"])
```

Pełny proces czyszczenia jest również w przykładach lekcji. Poniższe odpowiedzi używają gotowej kopii; odtwórz ją samodzielnie w notebooku lekcji przed porównaniem wyników.


## Sprzedaż na wynos

Wybierz wiersze z kanałem „na wynos”. Wyświetl kolumny `produkt`, `kanal`, `liczba_sztuk`, sortując po ilości malejąco.


```python
na_wynos = sprzedaz[sprzedaz["kanal"] == "na wynos"]
display(na_wynos[["produkt", "kanal", "liczba_sztuk"]].sort_values("liczba_sztuk", ascending=False).head(10))
```

**Interpretacja / kontrola:** W każdym pokazanym wierszu kanał to „na wynos”, a pierwsze ilości wynoszą 4.


## Kontrola i filtr z dwoma warunkami

Sprawdź braki i duplikaty w oczyszczonych danych. Wybierz sprzedaż co najmniej dwóch sztuk na wynos.


```python
print(sprzedaz.isna().sum())
print("Duplikaty:", sprzedaz.duplicated(subset="id_transakcji").sum())
wybrane = sprzedaz[(sprzedaz["liczba_sztuk"] >= 2) & (sprzedaz["kanal"] == "na wynos")]
display(wybrane.head())
```

**Interpretacja / kontrola:** Po pełnym czyszczeniu zostaje **236 wierszy**, bez braków i powtórzonych ID. Wybrany podzbiór spełnia oba warunki.
