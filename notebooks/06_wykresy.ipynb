| Jak porównują się kategorie? | Słupkowy |
| Jak wartość zmienia się w czasie? | Liniowy |

pandas przygotuje wykres, a matplotlib pomoże opisać i zapisać obraz. `ax` to obiekt z osiami wykresu, `fig` to cały obraz. Najpierw agreguj dane do poziomu pytania, potem rysuj.


## 1. Porównanie produktów: słupki

Zacznij oś wartości od zera. Długie nazwy łatwiej czytać na wykresie poziomym.


```python
przychody_produktow = sprzedaz.groupby("produkt")["przychod"].sum().sort_values()
ax = przychody_produktow.plot.barh(figsize=(8, 4), color="#387c9b")
ax.set(title="Przychód według produktu — styczeń–marzec 2026", xlabel="Przychód [zł]", ylabel="Produkt")
ax.set_xlim(left=0)
plt.tight_layout()
plt.show()
```

## 2. Zmiana w czasie: linia

Najpierw sumujemy przychody z tego samego dnia i porządkujemy daty. W zbiorze są obserwacje z każdego dnia okresu; nie musimy zgadywać, czy brak daty oznacza zero sprzedaży.


```python
dziennie = sprzedaz.groupby("data")["przychod"].sum().sort_index()
ax = dziennie.plot.line(figsize=(9, 4), color="#387c9b")
ax.set(title="Dzienny przychód kawiarni", xlabel="Data", ylabel="Przychód [zł]")
plt.tight_layout()
plt.show()
```

## 3. Rozkład: histogram

Histogram grupuje zakresy wartości liczbowych. `bins` określa liczbę przedziałów. To inny cel niż słupki porównujące nazwane kategorie.


```python
ax = sprzedaz["przychod"].plot.hist(bins=10, figsize=(7, 4), color="#387c9b", edgecolor="white")
ax.set(title="Rozkład wartości pozycji sprzedaży", xlabel="Wartość pozycji [zł]", ylabel="Liczba pozycji")
plt.tight_layout()
plt.show()
```

## 4. Zależność: punkty

Każdy punkt to jedna pozycja sprzedaży. `alpha` ustawia przezroczystość, więc nakładające się punkty są ciemniejsze. Cena i liczba sztuk są powtarzalne — punkty mogą się nakładać.


```python
ax = sprzedaz.plot.scatter(x="cena_jednostkowa", y="liczba_sztuk", alpha=0.25, figsize=(7, 4))
ax.set(title="Cena i liczba sztuk w pozycji sprzedaży", xlabel="Cena jednostkowa [zł]", ylabel="Liczba sztuk")
plt.tight_layout()
plt.show()
```

Zależność na wykresie nie dowodzi przyczyny. Wykres ceny i przychodu byłby dodatkowo częściowo skutkiem samego wzoru `przychód = cena × sztuki`. Na tych sztucznych danych nie wnioskujemy o prawdziwych zachowaniach klientów.
