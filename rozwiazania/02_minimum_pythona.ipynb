## Tydzień sprzedaży

Dla listy `[90, 120, 150, 100, 140]` oblicz sumę, średnią i najwyższy dzienny przychód.


```python
przychody = [90, 120, 150, 100, 140]
print("Suma:", sum(przychody))
print("Średnia:", sum(przychody) / len(przychody))
print("Maksimum:", max(przychody))
```

**Interpretacja / kontrola:** Odpowiednio: **600**, **120**, **150**.


## Sprawdź stan produktu

Utwórz słownik z nazwą produktu i stanem magazynowym. Jeżeli stan jest mniejszy niż 5, wyświetl „Zamów więcej”; w przeciwnym razie „Zapas wystarczy”. Sprawdź stan 3 i 8.


```python
produkt = {"nazwa": "Herbata", "stan": 3}
if produkt["stan"] < 5:
    print("Zamów więcej")
else:
    print("Zapas wystarczy")
```

**Interpretacja / kontrola:** Dla 3: „Zamów więcej”. Dla 8: „Zapas wystarczy”. Dla dokładnie 5 zapas też wystarczy według przyjętej reguły.


## Funkcja rabatu

Napisz funkcję `cena_po_rabacie(cena, procent)`, która zwraca cenę po obniżce. Wywołaj ją dla 20 zł i 15%.


```python
def cena_po_rabacie(cena, procent):
    return cena * (1 - procent / 100)

print(cena_po_rabacie(20, 15))
print(cena_po_rabacie(20, 0))
```

**Interpretacja / kontrola:** Dla 15% wynik to **17**, dla 0% to **20**.
