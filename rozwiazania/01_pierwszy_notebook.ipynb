## Sprzedaż herbaty

Sprzedano 8 herbat po 11 zł. Zapisz ilość i cenę w zmiennych, oblicz przychód i wyświetl go.


```python
liczba_herbat = 8
cena_herbaty = 11
przychod_herbaty = liczba_herbat * cena_herbaty
print(przychod_herbaty, "zł")
```

**Interpretacja / kontrola:** Wynik: **88 zł**. Czy po zmianie ilości na 10 wynik zmieni się na 110 zł?


## Cena po rabacie

Ciastko kosztuje 20 zł. Oblicz cenę po rabacie 15%.


```python
cena = 20
rabat_procent = 15
cena_po_rabacie = cena * (1 - rabat_procent / 100)
print(cena_po_rabacie)
```

**Interpretacja / kontrola:** Cena po rabacie: **17 zł**. Rabat wynosi 3 zł, a nie 15 zł.
