## Nowa umiejętność: różnica dat

Po konwersji możemy odejmować daty. `.dt.days` wydobywa liczbę dni z różnicy. `.dt` daje dostęp do właściwości kolumny dat, np. `.dt.month`. NaN w czasie wypożyczenia zostaje, gdy nie ma daty zwrotu.


```python
biblioteka["data_wypozyczenia"] = pd.to_datetime(biblioteka["data_wypozyczenia"], format="%Y-%m-%d")
biblioteka["data_zwrotu"] = pd.to_datetime(biblioteka["data_zwrotu"], format="%Y-%m-%d")
biblioteka["dni_wypozyczenia"] = (
    biblioteka["data_zwrotu"] - biblioteka["data_wypozyczenia"]
).dt.days
biblioteka[["data_wypozyczenia", "data_zwrotu", "dni_wypozyczenia"]].head(8)
```
