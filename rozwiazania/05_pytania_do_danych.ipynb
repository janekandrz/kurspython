To przykładowe odpowiedzi. Inny poprawny sposób jest równie dobry. Każde rozwiązanie uruchom po komórce przygotowującej dane poniżej.


```python
from pathlib import Path

# Notebook działa po uruchomieniu z katalogu kursu lub z jego podkatalogu.
KATALOG_KURSU = next(
    katalog for katalog in [Path.cwd(), *Path.cwd().parents]
    if (katalog / "dane" / "sprzedaz_kawiarni.csv").exists()
)
DANE = KATALOG_KURSU / "dane"
WYNIKI = KATALOG_KURSU / "wyniki"
WYNIKI.mkdir(exist_ok=True)

import pandas as pd
import matplotlib.pyplot as plt

sprzedaz = pd.read_csv(DANE / "sprzedaz_kawiarni_gotowa.csv", parse_dates=["data"])
sprzedaz.head()
```

## Ranking przychodu

Policz przychód dla każdego produktu, posortuj malejąco i wyświetl trzy najwyższe wyniki. Zapisz jedno zdanie o liderze, podając wartość.


```python
sprzedaz["przychod"] = sprzedaz["liczba_sztuk"] * sprzedaz["cena_jednostkowa"]
ranking = sprzedaz.groupby("produkt")["przychod"].sum().sort_values(ascending=False)
print(ranking.head(3))
print("Największy przychód:", ranking.index[0], "—", ranking.iloc[0], "zł")
```

**Interpretacja / kontrola:** Suma przychodów ze wszystkich produktów powinna być równa przychodowi całego zbioru. Lider przychodu nie musi być liderem liczby sztuk.


## Udział kanałów

Policz przychód kanałów „na miejscu” i „na wynos”, a potem ich udziały procentowe w całym przychodzie.


```python
sprzedaz["przychod"] = sprzedaz["liczba_sztuk"] * sprzedaz["cena_jednostkowa"]
kanaly = sprzedaz.groupby("kanal")["przychod"].sum()
udzialy = kanaly / kanaly.sum() * 100
print(kanaly)
print(udzialy.round(1))
print("Suma udziałów przed zaokrągleniem:", udzialy.sum())
```

**Interpretacja / kontrola:** Udziały przed zaokrągleniem sumują się do około **100%**. Wniosek powinien mówić o przychodzie, a nie o liczbie klientów.
