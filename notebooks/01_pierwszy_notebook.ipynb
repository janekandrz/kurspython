## 1. Python jako kalkulator

`*` oznacza mnożenie, `/` dzielenie. Najpierw zgadnij wynik.


```python
12 * 14
```

## 2. Zmienne: nazwy dla wartości

Znak `=` przypisuje wartość. Nazwy pomagają zrozumieć obliczenia. Python rozróżnia małe i wielkie litery.


```python
liczba_kaw = 12
cena_kawy = 15
przychod = liczba_kaw * cena_kawy
print(przychod)
```

## 4. Średnie i procenty

Nawiasy decydują o kolejności obliczeń. Udział procentowy to część podzielona przez całość i pomnożona przez 100.


```python
sprzedaz_poniedzialek = 120
sprzedaz_wtorek = 180
srednia = (sprzedaz_poniedzialek + sprzedaz_wtorek) / 2
udzial_poniedzialku = sprzedaz_poniedzialek / (sprzedaz_poniedzialek + sprzedaz_wtorek) * 100
print("Średni dzienny przychód:", srednia)
print("Udział poniedziałku:", udzial_poniedzialku, "%")
```

### Ćwiczenie: Cena po rabacie

Ciastko kosztuje 20 zł. Oblicz cenę po rabacie 15%.

<details><summary>Podpowiedź</summary>

Kwota rabatu to `20 * 15 / 100`. Odejmij ją od ceny.

</details>


```python
# Twoje rozwiązanie:
# Zastąp ten komentarz własnym kodem.
cena = 20
rabat = 0.15
nowaCena = cena * (1-rabat) 
print(nowaCena)
```

**Sprawdź swoją odpowiedź:** Cena po rabacie: **17 zł**. Rabat wynosi 3 zł, a nie 15 zł.


## 5. Błędy są informacją
