## 1. Lista: kilka wartości pod jedną nazwą

Indeks pierwszego elementu to `0`. `len()`, `sum()` i `max()` są funkcjami: przyjmują dane i zwracają wynik.


```python
przychody = [120, 180, 150, 210]
print("Pierwszy dzień:", przychody[0])
print("Liczba dni:", len(przychody))
print("Suma:", sum(przychody))
print("Średnia:", sum(przychody) / len(przychody))
```

## 2. Słownik: opis jednego rekordu

Słownik łączy klucze z wartościami. Do wartości sięgamy przez nazwę klucza w nawiasach kwadratowych. Lista słowników może opisać wiele rekordów.


```python
produkt = {"nazwa": "Latte", "cena": 16, "sztuki": 3}
print(produkt["nazwa"])
print(produkt["cena"] * produkt["sztuki"])

produkty = [
    {"nazwa": "Latte", "cena": 16},
    {"nazwa": "Herbata", "cena": 11},
]
print(produkty[1]["nazwa"])
```

## 3. Porównania i warunki

`==` porównuje, `=` przypisuje. `>`, `>=` i `<` porównują liczby. Wynik to `True` lub `False`. Po `if` oraz `else` jest dwukropek; kod wewnątrz ma wcięcie czterech spacji.


```python
przychod = 180
print(przychod >= 150)
if przychod >= 150:
    print("Cel osiągnięty")
else:
    print("Poniżej celu")
```

## 4. Pętla: powtórz działanie dla elementów

`for` bierze kolejno elementy listy. W pandas wiele podobnych operacji wykonamy od razu na całej kolumnie; teraz wystarczy rozumieć ten prosty zapis.


```python
for przychod in [120, 180, 150]:
    print("Dzienny przychód:", przychod)
```

## 5. Własna funkcja

`def` definiuje funkcję, parametry nazywają jej wejścia, a `return` zwraca wynik. Definicja nie wykonuje jeszcze obliczenia — potrzebne jest wywołanie. `print()` pokazuje wartość, ale nie zastępuje `return`.


```python
def wartosc_sprzedazy(cena, sztuki):
    return cena * sztuki

wynik = wartosc_sprzedazy(16, 3)
print(wynik)
print(wartosc_sprzedazy(cena=11, sztuki=2))
```
